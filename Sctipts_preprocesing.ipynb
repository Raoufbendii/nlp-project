```python
import os
import re
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from langdetect import detect
from arabic_reshaper import reshape
from bidi.algorithm import get_display
```

```python
# Download necessary NLTK data
nltk.download('punkt')
nltk.download('stopwords')
```

```python
# Define preprocessing function
def preprocess_text(text):
    # Detect language
    lang = detect(text)
    
    # Select stop words based on detected language
    if lang == 'ar':
        stop_words = set(stopwords.words('arabic'))
    else:
        stop_words = set(stopwords.words('english'))

    # Lowercase the text (if not Arabic)
    if lang != 'ar':
        text = text.lower()

    # Remove special characters, numbers, and extra spaces
    text = re.sub(r'[^\w\s]', '', text)
    text = re.sub(r'\d+', '', text)
    text = re.sub(r'\s+', ' ', text).strip()

    # Tokenize the text
    tokens = word_tokenize(text)

    # Remove stop words
    filtered_tokens = [word for word in tokens if word not in stop_words]

    # Normalize Arabic text (if applicable)
    if lang == 'ar':
        filtered_tokens = [reshape(word) for word in filtered_tokens]
        filtered_tokens = [get_display(word) for word in filtered_tokens]

    return ' '.join(filtered_tokens)
```

```python
# Directory containing the scripts
dir_path = 'Scripts'

# Process each file
for root, dirs, files in os.walk(dir_path):
    for file_name in files:
        # Skip log files or files with irrelevant extensions
        if file_name.endswith('.txt') or file_name.endswith('.py') or file_name.endswith('.ipynb'):
            file_path = os.path.join(root, file_name)
            
            with open(file_path, 'r', encoding='utf-8') as file:
                content = file.read()

            # Preprocess the text
            processed_text = preprocess_text(content)

            # Save the processed text to a new file
            output_path = os.path.join(root, f'processed_{file_name}')
            with open(output_path, 'w', encoding='utf-8') as output_file:
                output_file.write(processed_text)

print("Preprocessing complete. Processed files are saved in the respective folders.")
```
